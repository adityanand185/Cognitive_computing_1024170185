```python
#1
import numpy as np

arr = np.array([10,20,30,40,50])
print("Original array : ",arr)
print("After adding 2 : ",arr+2)
print("After multiplying by 3 : ",arr*3)
print("After dividing by 2 : ",arr/2)
```

```python
arr = np.array([1,2,3,6,4,5])
print("Original array : ",arr)
print("Reversed array : ",arr[::-1])

x = np.array([1,2,3,4,5,1,2,1,1,1])
value, counts = np.unique(x, return_counts=True)
max_count = np.max(counts)
most_frequent = value[counts == max_count]
print("Array x : ",x)
print("Most freqeuent value : ",most_frequent)
print("Freqeuency : ",max_count)

for value in most_frequent:
  print("Indices of",value,":",np.where(x == value)[0])

y = np.array([1,1,1,2,3,4,2,4,2,3,3])
value, counts = np.unique(y, return_counts=True)
max_count = np.max(counts)
most_frequent = value[counts==max_count]
print("Array y : ",y)
print("Most freqeuent value : ",most_frequent)
print("Freqeuency : ",max_count)

for value in most_frequent:
  print("Indices of",value,":",np.where(y == value)[0])

```

```python
#4

adi = np.linspace(10,100,25)
print("Array : ",adi)
print("Dimensions : ",adi.ndim)
print("Shape : ",adi.shape)
print("Total elements : ",adi.size)
print("Data type : ",adi.dtype)
print("Total bytes : ",adi.nbytes)
print("Transpose : ",adi.T)
reshaped = adi.reshape(1,25)
print("Transpose using reshape : ",reshaped.T)
print("Yes,we can use T attribute : ",adi.T)
```
