```python
#1. LIST
roll = input("Enter your roll number : ")
L = [int(digit)*10 for digit in roll]
```

```python
#1.i)
print("Original List : ",L)
```

```python
#1.ii)
L.append(100)
print("After append : ",L)
#appends adds 100 at the end of the list
L.insert(2,50)
print("After insert : ",L)
#insert adds 50 at position 2 and shift the elements to the right
```

```python
#1.iii)
L.remove(10)
print("After removing : ",L)
L.pop()
print("After popping : ",L)
```

```python
#1.iv)
L.sort()
print("After sorting in ascending : ",L)
L.sort(reverse = True)
print("After sorting in descending : ",L)
```

```python
#1.v)
print("First three elements : ",L[:3])
print("Last three elements : ",L[-3:])
```

```python
#1.vi)
average = sum(L)/len(L)
new_list = [x for x in L if x>average]
print("Average : ",average)
print("Element greater than average : ",new_list)
```

```python
#2 TUPLE
scores = tuple(L[:8])
```

```python
#2.i)
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_index = scores.index(lowest)
print("Highest score : ",highest," at index ",highest_index)
print("Lowest score : ",lowest," at index ",lowest_index)
```

```python
#2.ii)
reversed_scores = list(reversed(scores))
print("Reversed List : ",reversed_scores)
```

```python
#2.iii)
user_input = int(input("Enter a score to search: "))
if user_input in scores:
  print("First occurence index : ",scores.index(user_input))
else:
  print("Score not found")
```

```python
#2.iv)
try:
  scores[0] = 100
except TypeError as e:
  print("Error : ",e)
```

```python
#2.v)
first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
#3
import random
roll = int(input("Enter your roll number : "))
random.seed(roll)
```

```python
#3.i)
numbers = [random.randint(100,900) for i in range(100)]
print("Random numbers:",numbers)
```

```python
#3.ii)
odd_no = [x for x in numbers if x%2!=0]
print("Odd numbers:",odd_no)
print("Count of all odd numbers :",len(odd_no))
```

```python
#3.iii)
even_no = [x for x in numbers if x%2==0]
print("Even numbers:",even_no)
print("Count of all even numbers :",len(even_no))
```

```python
#3.iv)
def is_prime(n):
  if n<2:
    return False
  for i in range(2,int(n**0.5)+1):
    if n%i==0:
      return False
  return True

prime_no = [x for x in numbers if is_prime(x)]
print("Prime numbers:",prime_no)


prime_numbers = [x for x in numbers if is_prime(x)]
print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

```python
#3.v)
most_frequent = max(set(numbers),key=numbers.count)
frequency = numbers.count(most_frequent)
print("Most frequent number:",most_frequent)
print("Frequency:",frequency)
```

```python
#4
digits = [int(d) for d in str(roll)]
print("Roll number digits:",digits)
digits = digits[:8]
A= {digit*7 for digit in digits}
B= {digit*9 for digit in digits}
print("A:",A)
print("B:",B)
```

```python
#4.i)
union = A.union(B)
print("Union of A and B : ",union)
```

```python
#4.ii)
intersection = A.intersection(B)
print("Intersection of A and B : ",intersection)
```

```python
#4.iii)
difference = A.difference(B)
print("Difference of A - B : ",difference)
difference = B.difference(A)
print("Difference of B - A : ",difference)
```

```python
#4.iv)
symmetric_differnce = A.symmetric_difference(B)
print("The symemetrix difference of A and B :",symmetric_differnce)
```

```python
#4.v)
subset = A.issubset(B)
print("A is a Subset of B :",subset)
superset = B.issuperset(A)
print("B is a superset of B : ",superset)
```

```python
#4.vi)
X = int(input("Enter a value X to remove from Set A : "))
A.discard(X)
print("Set A after discarding : ",A)
```

```python
#5
my_dict = {
"name": "Aditya",
"roll_no": "1024170185",
"branch": "COPC",
"age": 22,
"city": "Bongaigaon"
}
print("Original Dictionary : ",my_dict)
```

```python
#5.i)
my_dict["location"] = my_dict.pop("city")
print("after naming city to location : ",my_dict)
```

```python
#5.ii)
my_dict["CGPA"] = 67
print("After adding CGPA : ",my_dict)

```

```python
#5.iii)
my_dict["age"] = my_dict["age"] + 1
print("After increasing by 1 : ",my_dict)
```

```python
#5.iv)
dict_pop = my_dict.copy()
dict_del = my_dict.copy()

removed_branch = dict_pop.pop("branch")

del dict_del["branch"]

print("\nUsing pop():",dict_pop)
print("Using del:",dict_del)
```

```python
#5.v)
print("Key-value pairs: ")
for key, value in my_dict.items():
    print(key, "->", value)
```

```python
#5.vi)
print("After adding CGPA :",my_dict)
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key not present.")
```
