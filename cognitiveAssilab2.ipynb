```python
#6.1
def odd(n):
  total = 0
  for i in range(1,n+1,2):
    if i%2!=0:
      total += i
  return total
n = int(input("Enter a number :"))
print("Sum of odd numbers from 1 to n ",n,"=",odd(n))
```

```python
#6.2
def is_prime(n):
  if n < 2:
    return False
  for i in range(2, int(n**0.5) + 1):
    if n % i == 0:
      return False
  return True
def sum_primes(num):
    total = 0
    for i in range(2, num + 1):
        if is_prime(i):
            total += i
    return total

num = int(input("Enter the value of n: "))
print("Sum of prime numbers from 1 to", num, "=", sum_primes(num))
```
