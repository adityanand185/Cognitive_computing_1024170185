```python
#6
import pandas as pd

data = {
    "Employee_ID": [101,102,103,104,105],
    "Name": ["Alice","Bob","Charlie","Diana","Edward"],
    "Department": ["HR","IT","IT","Marketing","Sales"],
    "Age": [29,34,41,28,38],
    "Salary": [50000,70000,65000,55000,60000],
    "Years_of_Experience": [4,8,10,3,12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female","Male","Male","Female","Male"],
    "Bonus": [5000,7000,6000,4500,5000],
    "Rating": [4.5,4.0,3.8,4.7,3.5]
}
df = pd.DataFrame(data)
print(df)

df.to_csv("employees.csv",index=False)
print("Employees dataset")
print(df)
```

```python
#5(a)
print("Shape:",df.shape)
```

```python
#5(b)
print("Summary: ",df.info())
```

```python
#5(c)
print("Descriptive Statistics : ",df.describe())
```

```python
#5(d)
print("First 5 rows : ",df.head(5))
print("Last 3 rows : ",df.tail(3))
```

```python
#5(e)
avg_sal = df["Salary"].mean()
print("Average Salary : ",avg_sal)

tot_bonus = df["Bonus"].sum()
print("Total Bonus : ",tot_bonus)

youngest_age = df["Age"].min()
print("Youngest Employee age : ",youngest_age)

highest_rating = df["Rating"].max()
print("Highest Performance rating : ",highest_rating)
```

```python
#5(f)
sorted_df = df.sort_values(by="Salary",ascending=False)
print("Salary in Descending Order : ",sorted_df)
```

```python
#5(g)
def performance_category(rating):
  if rating>=4.5:
    return "Excellent"

  elif rating>=4.0:
    return "Good"

  else:
    return "Average"

df["Performance"] = df["Rating"].apply(performance_category)
print("Performance cateogory : ",df[["Name", "Rating", "Performance"]])
```

```python
#5(h)

print("Mising values : ",df.isnull().sum())
```

```python
#5(i)
df = df.rename(columns={"Employee_ID": "ID"})
print("Renmaed Dataframe : ",df)
```

```python
#5(j)
experience = df[df["Years_of_Experience"]>5]
print("Employees with more than 5 years of experience : ",experience)

it_employees = df[df["Department"]=="IT"]
print("Employees in the IT department : ",it_employees)


```

```python
#5(k)
df["Tax"] = df["Salary"] * 0.10

print("\nK) Tax Column:")
print(df[["Name", "Salary", "Tax"]])
```

```python
#5(l)
df.to_csv("employees_modified.csv", index=False)

print("Modified DataFrame saved as employees_modified.csv")
```
