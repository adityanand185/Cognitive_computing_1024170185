```python
#1.
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
```

```python
#1.a
corrected_temperature = temperature + 2

print("Original temperatures:",temperature)
print("\nCorrected temperatures:",corrected_temperature)
```

```python
#1.b
fahrenheit = (9/5) * corrected_temperature + 32

print("\nTemperatures in Fahrenheit:",fahrenheit)
```

```python
#1.c
greater_than_32 = corrected_temperature[corrected_temperature > 32]

print("\nReadings greater than 32°C:",greater_than_32)
```

```python
#1.d
count = np.sum(corrected_temperature > 32)

print("\nNumber of readings exceeding 32°C:",count)
```

```python
#2
import numpy as np
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps matrix:",steps)
```

```python
#2.a
total_steps = np.sum(steps)

print("Total steps recorded:",total_steps)
```

```python
#2.b
mean_steps = np.mean(steps)

print("Mean number of steps:",mean_steps)
```

```python
#2.c
maximum = np.max(steps)
minimum = np.min(steps)

print("Maximum steps:",maximum)
print("Minimum steps:",minimum)
```

```python
#2.d
daily_total = np.sum(steps, axis=0)

print("Total steps for each day (axis=0):",daily_total)
```

```python
#2.e
user_total = np.sum(steps, axis=1)

print("Total steps for each user (axis=1):",user_total)
```

```python
#2.f
max_position = np.argmax(steps)
max_position_2d = np.unravel_index(np.argmax(steps), steps.shape)

print("Maximum step value:",steps[max_position_2d])

print("User/Day position:",max_position_2d)
```

```python
#3
import numpy as np
#a.
original = np.array([1, 2, 3, 4, 5, 6])

print("Original array:",original)
```

```python
#3.b
subset = original[1:5]

print("Subset:",subset)
```

```python
#3.c
subset[0] = 999

print("After modifying subset:")

print("Original:",original)

print("Subset:",subset)
```

```python
#3.d
copied = original[1:5].copy()

copied[0] = 500

print("After modifying copied array:")

print("Original:",original)

print("Copied:",copied)
```

```python
#3.e
matrix = np.arange(1, 13).reshape(3, 4)

print("3 x 4 matrix:",matrix)
```

```python
#3.f
first_row = matrix[0, :]
last_row = matrix[-1, :]
second_column = matrix[:, 1]
selected_elements = matrix[0:2, 1:3]

print("First row:",first_row)

print("Last row:",last_row)

print("Second column:",second_column)

print("Rows 1-2 and columns 2-3:",selected_elements)
```

```python
#3.g
flat_array = matrix.flatten()
ravel_array = matrix.ravel()

print("Flatten:",flat_array)

print("Ravel:",ravel_array)
```

```python
#3.h
ravel_array[0] = 100

print("After modifying ravel array:")
print("Ravel:",ravel_array)

print("Original matrix:",matrix)

print("Original matrix IS affected because ravel() usually returns a VIEW.")
```

```python
#3.i

flat_array[1] = 200

print("After modifying flatten array:")
print("Flatten:",flat_array)

print("Original matrix:",matrix)

print("Original matrix is NOT affected because flatten() returns a COPY.")
```

```python
#4
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
```

```python
#4.a
print("Shape of X:",X.shape)

print("Dimensions of X:",X.ndim)
```

```python
#4.b
X_T = X.T

print("Transpose of X (X.T):",X_T)

print("\nExplanation:")
print("Transpose changes rows into columns and columns into rows.")
```

```python
#4.c
XTX = X.T @ X

print("X.T @ X:",XTX)
```

```python
#4.d
XTX_inverse = np.linalg.inv(XTX)

print("Inverse of X.T @ X:",XTX_inverse)
```

```python
#4.e
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("OLS coefficients:",beta)
```

```python
#4.f
print("\nf. Interpretation of coefficients:")

print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

print("""
The coefficients represent the estimated effect of each feature
on the assistance score while keeping the other features constant.

Sleep coefficient:
A one-unit increase in sleep hours changes the predicted
assistance score by approximately the sleep coefficient.

Activity coefficient:
A one-unit increase in activity level changes the predicted
assistance score by approximately the activity coefficient.

Stress coefficient:
A one-unit increase in stress level changes the predicted
assistance score by approximately the stress coefficient.
""")
```

```python
#4.g
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("New user:",new_user)

print("Predicted assistance score:",predicted_score)
```
