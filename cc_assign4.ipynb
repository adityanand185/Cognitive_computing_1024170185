```python
#1
import pandas as pd
roll = input("Enter your roll number : ")

d1= int(roll[-2])
d2 = int(roll[-1])

fixed_entries = [
    {
        "question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge", "category": "billing"
    },
    {
        "question": "how to reset password", "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login", "category": "account"
    },
    {
        "question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time", "category": "general"
    },
    {
        "question": "how can i pay the fee", "answer" : "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee", "category": "billing"
    }
]

category1 = ["billing","account","general"][d1%3]
category2 = ["billing","account","general"][d2%3]

entry5 = {
    "question": "how do I update my registered mobile number",
    "answer": "Go to Account settings and update your registered mobile number,",
    "keywords": "mobile number update registered",
    "category": category1
}
entry6 = {
    "question": "how do I update my fee payment details",
    "answer": "You can update your fee payment details through the billing section.",
    "keywords": "fee payment billing update",
    "category": category2
}
df = pd.DataFrame(fixed_entries + [entry5,entry6])
print(df)
```

```python
#2
def score_query(query, df):
  result = []
  for index, row in df.iterrows():
    question_words = set(row["question"].lower().split())
    keyword_words = set(row["keywords"].lower().split())
    all_words = question_words.union(keyword_words)
    query_words = set(query.lower().split())
    score = len(query_words.intersection(all_words))
    if score > 0:
      result.append({
          "question": row["question"],
          "answer": row["answer"],
          "category": row["category"],
          "score": score
      })
  return result
query = input("Enter your query : ")
result = score_query(query, df)
print("Matching FAQs : ")
for result in result:
  print(f"Question : {result['question']}")
  print(f"Answer : {result['answer']}")
  print(f"Category : {result['category']}")
  print(f"Score : {result['score']}")
```

```python
#3
def same_category(category_name, df):
  return df[df["category"]==category_name]

category = input("enter category (billing/account/general) : ")
result = same_category(category, df)
print("\nFAQs in",category,"cateogry : ")
print(result)
```

```python
#4
index = 0
new_keyword = input("Enter a new keyword : ")
df.loc[index,"keywords"] = df.loc[index,"keywords"] + " " + new_keyword
df.to_csv("faq_data.csv",index=False)
print("Updated entry : ")
print(df.loc[index])
print("File saved as faq_data.csv")
```

```python
#6

def find_best_matches(query, df):
  query_words = set(query.lower().split())
  results = []
  for index, row in df.iterrows():
    question_words = set(row["question"].lower().split())
    keyword_words = set(row["keywords"].lower().split())
    all_words = question_words.union(keyword_words)
    score = len(query_words.intersection(all_words))
    results.append({
        "question": row["question"],
        "answer": row["answer"],
        "category": row["category"],
        "score": score
    })
  highest_score = max(result["score"] for result in results)
  best_matches = []
  for result in results:
    if result["score"] == highest_score:
      best_matches.append(result)
  return best_matches
query = input("Enter your query : ")
best_matches = find_best_matches(query, df)
print("Best matches : ")
for match in best_matches:
  print(f"Question : {match['question']}")
  print(f"Answer : {match['answer']}")
  print(f"Category : {match['category']}")
  print(f"Score : {match['score']}")
```
